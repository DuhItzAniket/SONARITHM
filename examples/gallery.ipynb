# SONARITHM examples gallery

Five synthetic datasets, each with a spec (`specs/`). This notebook compiles each one, prints its text description and shows it with an audio player. Run it from the `examples` directory, with `sonarithm` installed (`pip install sonarithm`).

```python
from pathlib import Path

import sonarithm as snr

print(snr.__version__)
SPECS = sorted(Path("specs").glob("ds-*.spec.json"))
[spec.name for spec in SPECS]
```

Each compile reports every default it applied, with its design-rule ID. The description is deterministic text, so a screen reader reads the same summary every time.

```python
gallery = {spec.name: snr.compile(snr.Spec.from_file(spec)) for spec in SPECS}
for name, compiled in gallery.items():
    print(compiled.describe())
```

Displaying a compiled spec shows its description and an audio player. Here, the three-region rainfall: three voices, left, centre and right.

```python
gallery["ds-003-regions.spec.json"]
```

Render to a file with its manifest (versions, hashes, loudness) and cue sidecar:

```python
manifest = gallery["ds-001-temperature.spec.json"].render("temperature.flac")
manifest["output"]["file"], manifest["output"]["integratedLufs"]
```

The same data from pandas: `quick` builds the default design (x along time, y as pitch).

```python
try:
    import pandas as pd
except ImportError:
    pd = None
if pd is not None:
    temps = pd.read_csv("data/ds-001-temperature.csv")
    quick = snr.quick(temps, x="day", y="temp", concept="temperature")
    print(quick.describe())
```
